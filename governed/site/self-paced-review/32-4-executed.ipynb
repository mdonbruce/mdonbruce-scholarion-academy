# Dependency-aware planning

Order a small plan only when prerequisites are met. Reject cycles rather than guessing an execution order.

Practice simulation only. No live AI, cloud execution, official grade or credential. Synthetic dictionaries are the entire dataset; no personal or externally licensed data. Python standard library only.

After testing, explain one rejected input, one remaining limitation and the human decision boundary.

```python
def order_plan(dependencies):
    pending={k:set(v) for k,v in dependencies.items()}
    done=[]
    while pending:
        ready=sorted(k for k,v in pending.items() if v <= set(done))
        if not ready:
            raise ValueError("blocked or cyclic plan")
        for key in ready:
            done.append(key)
            del pending[key]
    return done

assert order_plan({"evaluate":["build"],"build":["scope"],"scope":[]})==["scope","build","evaluate"]
for plan in ({"a":["b"],"b":["a"]},{"a":["missing"]}):
    try:
        order_plan(plan)
        raise AssertionError("blocked plan accepted")
    except ValueError:
        pass
print("All supplied practice checks passed")
```
