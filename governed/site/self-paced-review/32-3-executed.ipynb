# Typed tool boundaries

Validate a synthetic tool request before allowing it to reach a function. Treat booleans and missing values as invalid quantities.

Practice simulation only. No live AI, cloud execution, official grade or credential. Synthetic dictionaries are the entire dataset; no personal or externally licensed data. Python standard library only.

After testing, explain one rejected input, one remaining limitation and the human decision boundary.

```python
def validate_quantity(value):
    if type(value) is not int or not 1 <= value <= 10:
        raise ValueError("quantity rejected")
    return value

assert validate_quantity(3)==3
for value in (True, 0, 11, "3", None):
    try:
        validate_quantity(value)
        raise AssertionError("invalid input accepted")
    except ValueError:
        pass
print("All supplied practice checks passed")
```
