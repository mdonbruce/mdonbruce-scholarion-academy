# Approval-gated graph

Advance a synthetic workflow checkpoint only after explicit approval. A pending decision must preserve the review checkpoint.

Practice simulation only. No live AI, cloud execution, official grade or credential. Synthetic dictionaries are the entire dataset; no personal or externally licensed data. Python standard library only.

After testing, explain one rejected input, one remaining limitation and the human decision boundary.

```python
def advance(state, approved=False):
    if state == "draft": return "review"
    if state == "review": return "complete" if approved is True else "review"
    if state == "complete": return "complete"
    raise ValueError("unknown checkpoint")

assert advance("draft")=="review"
assert advance("review")=="review"
assert advance("review",True)=="complete"
assert advance("review","yes")=="review"
print("All supplied practice checks passed")
```
